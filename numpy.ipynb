```python
import numpy as np
import os
```

```python
arr = np.array([3, 1, 4, 1, 5])
# Crea una copia ordinata
ordinato = np.sort(arr)


print("Originale:", arr)
print("Ordinato:", ordinato)
```

```python
arr.sort()
print(arr)
```

```python
nomi = np.array(['Mario', 'Laura', 'Giuseppe'])
voti = np.array([7.5, 9.0, 8.0])
indice=np.argsort(voti)
print(indice)
for numero in indice:
    print(f"{nomi[numero]} : {voti[numero]}")
```

```python
studenti = np.array([
    ('Mario', 25, 8.5),
    ('Laura', 23, 9.0),
    ('Giuseppe', 27, 7.5)
], dtype=[('nome', 'U10'), ('età', 'i4'), ('voto', 'f4')])
ordvoto=np.sort(studenti, order="voto")
print(ordvoto)
ordvoto1=np.sort(studenti, order=["età","voto"])
print(ordvoto1)
ordecrescente=np.sort(studenti, order="voto")[::-1]
print(ordecrescente)
```

```python
matrice = np.array([[3, 1, 4],
                    [1, 5, 9],
                    [2, 6, 5]])
matord=np.sort(matrice, axis=0)
matord1=np.sort(matrice, axis=1)
print(matord)
print(matord1)
```

```python
# Array ordinato
arr = np.array([1, 3, 5, 7, 9])


# Dove dovrei inserire 6?
idx = np.searchsorted(arr, 6)
print(f"Inserisci 6 alla posizione: {idx}")


# Verifica
arr_nuovo = np.insert(arr, idx, 6)
print(f"Nuovo array: {arr_nuovo}")
print()


# Inserire più valori
valori = [2, 6, 8]
for val in valori:
    idx = np.searchsorted(arr, val)
    print(f"Valore {val} dovrebbe andare alla posizione {idx}")
```

```python
arr = np.array([1, 3, 5, 7, 9])
min=np.argmin(arr)
max=np.argmax(arr)
print(min)
print(max)
```

```python
matrice = np.array([[1, 0, 2],
                    [6, 3, 0],
                    [4, 0, 5]])
print(np.argmin(matrice, axis=0))
print(np.argmin(matrice, axis=1))
```

```python
arr5 = np.array([0, 1, 0, 3, 0, 5])
indici=np.nonzero(arr5)
print(indici)
```

```python
# Accedi ai valori non-zero
valori_nonzero = arr5[np.nonzero(arr5)]
print(f"Valori non-zero: {valori_nonzero}")  # [1 3 5]
print()


# Con matrici
matrice = np.array([[1, 0, 2],
                    [0, 3, 0],
                    [4, 0, 5]])


print("Posizioni degli elementi non-zero (matrice):")
indici_mat = np.nonzero(matrice)
for riga, col in zip(indici_mat[0], indici_mat[1]):
    print(f"({riga}, {col}): {matrice[riga, col]}")
```

```python
dati = np.array([[1.234, 5.678, 9.012],
                 [2.345, 6.789, 0.123]])
# CSV semplice
np.savetxt('dati.csv', dati,
           delimiter=' , ',
           fmt='%.3f',
           header='Col1, Col2, Col3',
           comments='tabella\n')


print("CSV salvato!")
```

```python
# Caricamento con np.loadtxt()
dati_csv = np.loadtxt('dati.csv', delimiter=',', skiprows=2)
print(f"Dati caricati:\n{dati_csv}")
print()
x, y = np.loadtxt('dati.csv', delimiter=',', usecols=(0, 1), unpack=True, skiprows=2)
print(f"Colonna X: {x}")
print(f"Colonna Y: {y}")
```
