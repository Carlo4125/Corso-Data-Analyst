```python
df_amazon["order_date"]=pd.to_datetime(df_amazon["order_date"])
```

```python
df_amazon["mese_anno"]=df_amazon["order_date"].dt.to_period("M")

annual_sales=df_amazon.groupby("mese_anno").agg(
    Total_revenue=("total_revenue","sum"),
    Total_sales=("order_id","count")
)


plt.figure(figsize=(10,5))
annual_sales["Total_revenue"].plot(kind='line', marker="o", color="black", linewidth=2)
plt.title("VENDITE MENSILI")
plt.xlabel("Anno.Mese")
plt.ylabel("Vendite")
plt.xticks(rotation=45, ha="right")
plt.grid(axis='y', alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
df_febbraio=df_amazon[df_amazon["mese_anno"]=="2023-02"].copy()


df_febbraio["day"]=df_febbraio["order_date"].dt.day



feb_sales=df_febbraio.groupby("day").agg(
    Totale=("total_revenue","sum"),
    Quantita=("order_id","count")
)

plt.figure(figsize=(10,5))
feb_sales["Totale"].plot(kind='line', marker="o", color="black", linewidth=2)
plt.title("VENDITE MESE FEBBRAIO")
plt.xlabel("Giorno")
plt.ylabel("Vendite")
plt.xticks(rotation=45, ha="right")
plt.grid(axis='y', alpha=0.3)
plt.tight_layout()
plt.show()

```

```python
import matplotlib.pyplot as plt

# Lista degli sconti e dei colori (puoi personalizzare)
sconti = [0, 5, 10, 15, 20, 30]
colori = ["black", "red", "blue", "green", "orange", "purple"]

plt.figure(figsize=(12,6))

for sconto, colore in zip(sconti, colori):
    # filtro il DataFrame e creo la colonna anno_mese
    df_tmp = df_amazon[df_amazon["discount_percent"] == sconto].copy()
    df_tmp["anno_mese"] = df_tmp["order_date"].dt.to_period("M")
    
    # raggruppo per mese e sommo i ricavi
    monthly_rev = df_tmp.groupby("anno_mese")["total_revenue"].sum()
    
    # plotta la linea
    plt.plot(
        monthly_rev.index.astype(str),  # converti i periodi in stringhe per l'asse X
        monthly_rev.values,
        marker='o',
        color=colore,
        label=f"{sconto}% sconto"
    )

plt.title("Vendite mensili per livello di sconto")
plt.xlabel("Anno-Mese")
plt.ylabel("Totale ricavi")
plt.xticks(rotation=45, ha="right")
plt.grid(axis='y', alpha=0.3)
plt.legend()
plt.tight_layout()
plt.show()
```
